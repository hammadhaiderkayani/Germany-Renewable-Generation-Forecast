```python
from pathlib import Path
import hashlib

base = SOURCE_ROOT / "germany_weather_data"

solar_files = [
    base / "renewable_models" / "solar_model.pkl",
    base / "calendar_models" / "solar_mw_calendar_model.pkl",
    base / "trained_models" / "solar_model.joblib"
]

for path in solar_files:
    if path.exists():
        sha = hashlib.sha256(path.read_bytes()).hexdigest()

        print("\nFile:", path.name)
        print("Folder:", path.parent.name)
        print("Size:", round(path.stat().st_size / 1024, 2), "KB")
        print("SHA256:", sha)
    else:
        print("\nNOT FOUND:", path)
```

```python
import joblib

model_paths = {
    "renewable_models": base / "renewable_models" / "solar_model.pkl",
    "trained_models": base / "trained_models" / "solar_model.joblib"
}

for name, path in model_paths.items():
    model = joblib.load(path)

    print("\n", "=" * 55)
    print(name.upper())
    print("=" * 55)

    print("Model class:", type(model))
    print("Number of estimators:",
          getattr(model, "n_estimators", "Not available"))
    print("Number of features:",
          getattr(model, "n_features_in_", "Not available"))
```

```python
for name, path in model_paths.items():
    model = joblib.load(path)

    print("\n", "=" * 55)
    print(name.upper())
    print("=" * 55)

    print("Features:", model.n_features_in_)
    print("Iterations:", model.n_iter_)
    print("Maximum iterations:", model.max_iter)
    print("Learning rate:", model.learning_rate)
    print("Maximum leaf nodes:", model.max_leaf_nodes)
    print("Maximum depth:", model.max_depth)
    print("L2 regularization:", model.l2_regularization)
```

```python
import hashlib

repo_solar = PROJECT_ROOT / "models" / "Solar" / "solar_model.pkl"
original_solar = base / "renewable_models" / "solar_model.pkl"

def file_hash(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()

print("GitHub project model:")
print(file_hash(repo_solar))

print("\nOriginal model:")
print(file_hash(original_solar))

print("\nAre they identical?")
print(file_hash(repo_solar) == file_hash(original_solar))
```
