- Identify potential issues before feature engineering and model development.

**Geographical scope:** Germany  
**Forecasting targets:** Onshore wind, offshore wind, and solar generation (MW).

```python
import pandas as pd
import joblib
from pathlib import Path

# --------------------------------------------------
# 1. Check exact feature names in the GitHub models
# --------------------------------------------------
print("=" * 65)
print("MODEL FEATURE CHECK")
print("=" * 65)

solar_features = joblib.load(root / "models" / "Shared" / "feature_columns.pkl")
wind_models = joblib.load(root / "models" / "Wind" / "final_wind_models.pkl")

print("Solar features:", len(solar_features))
print("Solar first 5:", solar_features[:5])

for name, model in wind_models.items():
    if hasattr(model, "booster_"):
        features = model.booster_.feature_name()
    elif hasattr(model, "feature_names_in_"):
        features = list(model.feature_names_in_)
    else:
        features = None

    print(f"\n{name} model:")
    print("Expected features:", model.n_features_in_)
    if features:
        print("Feature names available:", len(features))
        print("First 8:", features[:8])
        print("Last 8:", features[-8:])
    else:
        print("Feature names not available directly.")

# --------------------------------------------------
# 2. Find actual generation data files
# --------------------------------------------------
print("\n" + "=" * 65)
print("SEARCHING FOR ACTUAL GENERATION DATA")
print("=" * 65)

search_root = base.parent
keywords = ["generation", "renewable", "smard", "actual"]
excluded = ["weather", "feature_importance", "diagnostic", "prediction"]

candidates = []

for path in search_root.rglob("*"):
    if not path.is_file() or path.suffix.lower() not in [".csv", ".parquet"]:
        continue

    name = path.name.lower()
    if any(word in name for word in keywords) and not any(word in name for word in excluded):
        if ".git" not in path.parts:
            candidates.append(path)

for path in sorted(candidates):
    print("\nFile:", path)
    try:
        if path.suffix.lower() == ".csv":
            sample = pd.read_csv(path, nrows=2)
            print("Columns:", sample.columns.tolist())
        else:
            import pyarrow.parquet as pq
            metadata = pq.ParquetFile(path)
            print("Rows:", metadata.metadata.num_rows)
            print("Columns:", metadata.schema.names)
    except Exception as e:
        print("Inspection error:", e)

print("\nCandidate files found:", len(candidates))
```

```python
import pandas as pd
import numpy as np
import joblib

test_date = "2024-06-02"

# Load saved model artifacts
solar_model = joblib.load(root / "models" / "Solar" / "solar_model.pkl")
solar_features = joblib.load(root / "models" / "Shared" / "feature_columns.pkl")

wind_models = joblib.load(root / "models" / "Wind" / "final_wind_models.pkl")
wind_calibrators = joblib.load(root / "models" / "Wind" / "final_wind_calibrators.pkl")

# Extract exact feature names from trained wind models
def get_model_features(model):
    if hasattr(model, "feature_name_"):
        return list(model.feature_name_)
    if hasattr(model, "feature_names_in_"):
        return list(model.feature_names_in_)
    return list(model.booster_.feature_name())

offshore_features = get_model_features(wind_models["offshore"])
onshore_features = get_model_features(wind_models["onshore"])

# Load weather archives
weather_17 = pd.read_csv(base / "germany_archived_weather_17_locations_experimental.csv")
weather_22 = pd.read_csv(base / "germany_archived_weather_22_locations_offshore_experimental.csv")

for df in [weather_17, weather_22]:
    df["time"] = pd.to_datetime(df["time"], utc=True, errors="coerce")
    df.drop_duplicates(["time", "location"], inplace=True)

# Convert long weather data into wide hourly features
def make_wide_weather(df):
    day = df[df["time"].dt.strftime("%Y-%m-%d") == test_date].copy()

    direction = day.pivot(index="time", columns="location", values="wind_direction_100m")
    direction_sin = np.sin(np.deg2rad(direction))
    direction_cos = np.cos(np.deg2rad(direction))
    direction_sin.columns = [f"wind_direction_sin_{c}" for c in direction_sin.columns]
    direction_cos.columns = [f"wind_direction_cos_{c}" for c in direction_cos.columns]

    wide = {}
    for variable in [
        "wind_speed_100m", "temperature_2m",
        "shortwave_radiation", "direct_radiation",
        "diffuse_radiation", "cloud_cover"
    ]:
        part = day.pivot(index="time", columns="location", values=variable)
        part.columns = [f"{variable}_{c}" for c in part.columns]
        wide[variable] = part

    result = pd.concat([*wide.values(), direction_sin, direction_cos], axis=1)
    result = result.sort_index()

    time_frame = pd.DataFrame({"time": result.index}, index=result.index)
    hour = time_frame["time"].dt.hour
    doy = time_frame["time"].dt.dayofyear
    result["hour_sin"] = np.sin(2 * np.pi * hour / 24)
    result["hour_cos"] = np.cos(2 * np.pi * hour / 24)
    result["day_of_year_sin"] = np.sin(2 * np.pi * doy / 365.25)
    result["day_of_year_cos"] = np.cos(2 * np.pi * doy / 365.25)

    return result

X_solar = make_wide_weather(weather_17)
X_onshore = make_wide_weather(weather_17)
X_offshore = make_wide_weather(weather_22)

# Report exact feature availability
for name, X, features in [
    ("Solar", X_solar, solar_features),
    ("Onshore", X_onshore, onshore_features),
    ("Offshore", X_offshore, offshore_features)
]:
    missing = [f for f in features if f not in X.columns]

    print("\n" + "=" * 65)
    print(name)
    print("Weather rows:", len(X))
    print("Available columns:", len(X.columns))
    print("Required model features:", len(features))
    print("Missing required features:", len(missing))
    if missing:
        print("Missing feature names:", missing)
    else:
        print("All model features are present.")
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error

# ============================================================
# JUNE 2024 — COMPLETE THREE-MODEL HISTORICAL BACKTEST
# ============================================================

start = pd.Timestamp("2024-06-01", tz="UTC")
end = pd.Timestamp("2024-06-30 23:00", tz="UTC")

variables = [
    "wind_speed_100m",
    "temperature_2m",
    "shortwave_radiation",
    "direct_radiation",
    "diffuse_radiation",
    "cloud_cover"
]

# 1. Prepare hourly weather features using the saved naming convention
def prepare_weather_features(df):
    df = df.copy()
    df["time"] = pd.to_datetime(df["time"], utc=True, errors="coerce")
    df = df[df["time"].between(start, end)]
    df = df.drop_duplicates(["time", "location"])

    result = pd.DataFrame(
        index=pd.DatetimeIndex(sorted(df["time"].unique()), name="time")
    )

    for variable in variables:
        part = df.pivot(
            index="time", columns="location", values=variable
        )
        part.columns = [f"{variable}_{loc}" for loc in part.columns]
        result = result.join(part)

    direction = df.pivot(
        index="time", columns="location", values="wind_direction_100m"
    )

    sin_part = np.sin(np.deg2rad(direction))
    cos_part = np.cos(np.deg2rad(direction))

    sin_part.columns = [f"wind_direction_sin_{loc}" for loc in sin_part.columns]
    cos_part.columns = [f"wind_direction_cos_{loc}" for loc in cos_part.columns]

    result = result.join(sin_part).join(cos_part)

    hour = result.index.hour
    doy = result.index.day_of_year

    result["hour_sin"] = np.sin(2 * np.pi * hour / 24)
    result["hour_cos"] = np.cos(2 * np.pi * hour / 24)
    result["day_of_year_sin"] = np.sin(2 * np.pi * doy / 365.25)
    result["day_of_year_cos"] = np.cos(2 * np.pi * doy / 365.25)

    return result.sort_index()


# 2. Prepare weather for all three models
X_solar_month = prepare_weather_features(weather17)
X_onshore_month = prepare_weather_features(weather17)
X_offshore_month = prepare_weather_features(weather22)

# Add North Sea 4 and 5 using the offshore model's prefix naming
offshore_source = weather22.copy()
offshore_source["time"] = pd.to_datetime(
    offshore_source["time"], utc=True, errors="coerce"
)
offshore_source = offshore_source[
    offshore_source["time"].between(start, end)
]

for loc in ["North_Sea_4", "North_Sea_5"]:
    loc_data = offshore_source[
        offshore_source["location"] == loc
    ].drop_duplicates("time").set_index("time")

    for variable in variables:
        X_offshore_month[f"{loc}_{variable}"] = loc_data[variable]

    radians = np.deg2rad(loc_data["wind_direction_100m"])
    X_offshore_month[f"{loc}_wind_direction_sin"] = np.sin(radians)
    X_offshore_month[f"{loc}_wind_direction_cos"] = np.cos(radians)

# 3. Select exact saved model features and order
X_solar_month = X_solar_month[solar_features]
X_onshore_month = X_onshore_month[onshore_features]
X_offshore_month = X_offshore_month[offshore_features]

# 4. Validate model inputs
for name, X in [
    ("Solar", X_solar_month),
    ("Onshore", X_onshore_month),
    ("Offshore", X_offshore_month)
]:
    print(
        f"{name}: {len(X)} rows, {X.shape[1]} features, "
        f"{X.isna().sum().sum()} missing values"
    )

    if len(X) != 720 or not np.isfinite(X.to_numpy(dtype=float)).all():
        raise ValueError(f"{name}: incomplete or invalid model inputs")

# 5. Run saved models
solar_prediction = solar_model.predict(X_solar_month)

onshore_raw = wind_models["onshore"].predict(X_onshore_month)
offshore_raw = wind_models["offshore"].predict(X_offshore_month)

onshore_prediction = wind_calibrators["onshore"].predict(
    np.asarray(onshore_raw).reshape(-1, 1)
)

offshore_prediction = wind_calibrators["offshore"].predict(
    np.asarray(offshore_raw).reshape(-1, 1)
)

# 6. Load and align actual generation
actual_month = generation.copy()
actual_month["time"] = pd.to_datetime(
    actual_month["time"], utc=True, errors="coerce"
)
actual_month = actual_month[
    actual_month["time"].between(start, end)
].drop_duplicates("time").set_index("time").sort_index()

results_month = pd.DataFrame(index=X_solar_month.index)

results_month["solar_actual"] = actual_month["solar_mw"].reindex(
    results_month.index
)
results_month["solar_predicted"] = solar_prediction

results_month["onshore_actual"] = actual_month["onshore_wind_mw"].reindex(
    results_month.index
)
results_month["onshore_predicted"] = onshore_prediction

results_month["offshore_actual"] = actual_month["offshore_wind_mw"].reindex(
    results_month.index
)
results_month["offshore_predicted"] = offshore_prediction

if len(actual_month) != 720 or results_month.isna().any().any():
    raise ValueError("Actual generation does not align with all 720 forecast hours.")

# 7. Calculate monthly metrics
def evaluate_month(actual, predicted):
    error = np.asarray(predicted) - np.asarray(actual)

    return {
        "MAE_MW": mean_absolute_error(actual, predicted),
        "RMSE_MW": np.sqrt(mean_squared_error(actual, predicted)),
        "Bias_MW": np.mean(error),
        "WAPE_percent": (
            np.sum(np.abs(error)) / np.sum(np.abs(actual)) * 100
        )
    }

monthly_metrics = pd.DataFrame({
    "Solar": evaluate_month(
        results_month["solar_actual"], results_month["solar_predicted"]
    ),
    "Onshore": evaluate_month(
        results_month["onshore_actual"], results_month["onshore_predicted"]
    ),
    "Offshore": evaluate_month(
        results_month["offshore_actual"], results_month["offshore_predicted"]
    )
}).T

# 8. Daily MAE summary to see consistency across the month
daily_errors = {}

for technology in ["solar", "onshore", "offshore"]:
    absolute_error = (
        results_month[f"{technology}_predicted"]
        - results_month[f"{technology}_actual"]
    ).abs()

    daily_errors[technology] = absolute_error.groupby(
        results_month.index.date
    ).mean()

daily_mae = pd.DataFrame(daily_errors)
daily_mae.index.name = "date"

# 9. Display results
print("\n" + "=" * 75)
print("JUNE 2024 — MONTHLY MODEL PERFORMANCE")
print("=" * 75)
print(monthly_metrics.round(2).to_string())

print("\n" + "=" * 75)
print("DAILY MAE SUMMARY — MW")
print("=" * 75)
print(daily_mae.describe().round(2).to_string())

print("\nDays evaluated:", len(daily_mae))
print("Hourly observations:", len(results_month))
print("All three models completed successfully.")
```

```python
# Compare daily WAPE for all three models

daily_results = []

for technology in ["solar", "onshore", "offshore"]:
    actual_col = f"{technology}_actual"
    predicted_col = f"{technology}_predicted"

    for date, group in results_month.groupby(results_month.index.date):
        actual = group[actual_col]
        predicted = group[predicted_col]

        wape = (
            (predicted - actual).abs().sum()
            / actual.abs().sum()
            * 100
        )

        daily_results.append({
            "date": date,
            "technology": technology,
            "WAPE_percent": wape
        })

daily_wape = pd.DataFrame(daily_results)

print("AVERAGE DAILY WAPE")
print(
    daily_wape.groupby("technology")["WAPE_percent"]
    .agg(["mean", "median", "min", "max"])
    .round(2)
)

print("\nWORST 5 DAYS PER TECHNOLOGY")

for technology in ["solar", "onshore", "offshore"]:
    print(f"\n{technology.upper()}")
    print(
        daily_wape[daily_wape["technology"] == technology]
        .nlargest(5, "WAPE_percent")
        .round(2)
        .to_string(index=False)
    )
```
